```python
# Import the function used to divide the dataset into training and testing sets
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing subsets
# 80% of the data is used for training and 20% for testing
# random_state=45 ensures the same split is obtained every time
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.3,
    random_state=45
)
```

```python
print("X_train:")
print(x_train)

print("\nX_test:")
print(x_test)

print("\nY_train:")
print(y_train)

print("\nY_test:")
print(y_test)
```

```python
print("X_train shape:", x_train.shape)
print("X_test shape:", x_test.shape)

print("Y_train shape:", y_train.shape)
print("Y_test shape:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

sc_x = StandardScaler()

x_train = sc_x.fit_transform(x_train)

x_test = sc_x.transform(x_test)
x_test
```

```python
# Import the Decision Tree classification algorithm
from sklearn.tree import DecisionTreeClassifier

# Create the Decision Tree classifier
# criterion='entropy' uses information gain to select the best split
# random_state=0 ensures reproducible results
classifier = DecisionTreeClassifier(
    criterion='entropy',
    random_state=0
)

# Train the Decision Tree model using the training data
classifier.fit(x_train, y_train)
```

```python
y_pred=classifier.predict(x_test)
y_pred
```

```python
y_test
```

```python
# Import the confusion matrix evaluation metric
from sklearn.metrics import confusion_matrix

# Create the confusion matrix by comparing
# actual values with predicted values

cm = confusion_matrix(y_test, y_pred)
cm
```

```python
Accuracy=(cm[0][0]+cm[1][1])/(cm[0][0]+cm[0][1]+cm[1][0]+cm[1][1])
Accuracy
```

```python
from sklearn import metrics

prediction=metrics.accuracy_score(y_test,y_pred)
print("Accuracy: ",prediction *100,"%")
```

```python
# Import classification_report to calculate
# precision, recall, F1-score, and support
from sklearn.metrics import classification_report

# Display the classification performance of the model
print(classification_report(y_test, y_pred))
```
