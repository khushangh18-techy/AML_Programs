5. Implementation of the quicksort algorithm in Python. [3,6,8,10,1,2,1]

```python
def quicksort(arr):
    # Base case: lists with 0 or 1 element are already sorted
    if len(arr) <= 1:
        return arr
    
    # Choosing the middle element as the pivot
    pivot = arr[len(arr) // 2]
    
    # Partitioning the list into three parts
    left = [x for x in arr if x < pivot]    # Elements smaller than pivot
    middle = [x for x in arr if x == pivot] # Elements equal to pivot
    right = [x for x in arr if x > pivot]  # Elements greater than pivot
    
    # Recursively sort the sub-arrays and combine them
    return quicksort(left) + middle + quicksort(right)

# Test the function with the given list
numbers = [3, 6, 8, 10, 1, 2, 1]
sorted_numbers = quicksort(numbers)

print("Original list:", numbers)
print("Sorted list:  ", sorted_numbers)
```

6. computes square numbers using mapping function/ lambda function

```python

numbers = [1, 2, 3, 4, 5]

squared_numbers = list(map(lambda x: x ** 2, numbers))

print("Original numbers:", numbers)
print("Squared numbers: ", squared_numbers)
```

7. Compute square of numbers using list comprehension technique.

```python
numbers = [1, 2, 3, 4, 5]

# Syntax: [expression for item in iterable]
squared_numbers = [x ** 2 for x in numbers]

print("Original numbers:", numbers)
print("Squared numbers: ", squared_numbers)
```

8. Compute square of even numbers using  using list comprehension technique.

```python
# 1. Define a list of numbers
numbers =[2,4,8]

# 2. List comprehension with a conditional if statement
# Syntax: [expression for item in iterable if condition]
even_squares = [x ** 2 for x in numbers if x % 2 == 0]

# 3. Print the result
print("Original numbers:", numbers)
print("Even squares:    ", even_squares)
```
