```python
#split the independent veriables
#loc uses the labels
#iloc uses the integer positions (no. starting from 0 to length of the data)
df
x = df.iloc[:, :-1].values
x
```

```python
#split dependent variable
y = df.iloc[:, 3].values
y
```

```python
# taking care of missing data
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer = imputer.fit(x[:, 1:3])

x[:, 1:3] = imputer.transform(x[:, 1:3])
x
```

```python
# encoding the categorical data
#encoding the independent veriable
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
labelencoder_X = LabelEncoder()
x[:, 0] = labelencoder_X.fit_transform(x[:, 0])
print(x)
```

```python
# encoding the independent variable
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(
    [('country', OneHotEncoder(), [0])],
    remainder='passthrough'
)

x = ct.fit_transform(x)
x = x.astype(int)


```

```python
#split the data into the training set and Test set
from sklearn.model_selection import train_test_split
x_train, x_test, y_train,y_test = train_test_split(x, y, test_size=0.2, random_state=45)
```

```python
#Feature Scaling
#sc_X = (x[1] - mean) / standard deviation
from sklearn.preprocessing import StandardScaler
sc_X = StandardScaler()
x_train = sc_X.fit_transform(x_train)
x_test = sc_X.fit_transform(x_test)
```
