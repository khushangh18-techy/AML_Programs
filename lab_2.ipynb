```python
num = np.matrix([[11, 121, 130],
                [613, 34, 13],
                [514, 50, 16]])
```

```python
num
```

```python
type(num)
```

```python
print("Main Diagonal elements : \n", np.diag(num))
print("\n Diagonal elements above main diagonal : \n", np.diag(num, 1))
print("\n Diagonal elements below main diagonal : \n", np.diag(num, -1))
print("\n Diagonal elements one step above main diagonal : \n", np.diag(num,2))
print("\n Diagonal elements one step below main diagonal : \n", np.diag(num,-2))
```
