```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('C:/Users/MSIS/Downloads/AML/MLLabprojects/data/Social_Network_Ads.csv')

x = df[['Age', 'EstimatedSalary']]
y = df['Purchased']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 42, stratify = y)

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)
x_test
```

```python
#fitting decision tree classification to the training set
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 42)
classifier.fit(x_train, y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'gini', random_state = 0)
classifier.fit(x_train, y_train)
```

```python
#predicting the test set results
y_pred = classifier.predict(x_test)
y_pred
```

```python
from sklearn import metrics
#model accuracy, how often is the classifier correct?
prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy:",prediction*100, '%')
```

K-Fold Cross Validation, K=3

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.tree import DecisionTreeClassifier

skfold = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

classifier = DecisionTreeClassifier(
    criterion='entropy',
    random_state=42
)

scores = cross_val_score(
    classifier,
    x,
    y,
    cv=skfold,
    scoring='accuracy'
)

print("K = 3")
print("Fold Accuracies:", scores)
print("Mean Accuracy:", scores.mean())
```

Checking for other paramenters in decision tree

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.tree import DecisionTreeClassifier

# Load dataset
df = pd.read_csv(
    'C:/Users/MSIS/Downloads/AML/MLLabprojects/data/Social_Network_Ads.csv'
)

# Select features and target
x = df[['Age', 'EstimatedSalary']]
y = df['Purchased']

# 5-Fold Cross Validation
skfold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# =========================================================
# 1. MAX_DEPTH COMPARISON
# =========================================================

depths = [1, 2, 3, 4, 5, 6, 7, None]

depth_results = []

for depth in depths:

    classifier = DecisionTreeClassifier(
        criterion='entropy',
        max_depth=depth,
        min_samples_split=2,
        min_samples_leaf=1,
        random_state=42
    )

    scores = cross_val_score(
        classifier,
        x,
        y,
        cv=skfold,
        scoring='accuracy'
    )

    depth_results.append([
        depth,
        scores.mean()
    ])

depth_df = pd.DataFrame(
    depth_results,
    columns=['Max Depth', 'Mean Accuracy']
)

print("\nMAX_DEPTH COMPARISON")
print(depth_df)

best_depth_index = depth_df['Mean Accuracy'].idxmax()
best_depth = depths[best_depth_index]

print("\nBest Max Depth:", best_depth)
print("Best Mean Accuracy:",
      depth_df.loc[best_depth_index, 'Mean Accuracy'])


# =========================================================
# 2. MIN_SAMPLES_SPLIT COMPARISON
# =========================================================

split_values = [2, 5, 10, 15, 20]

split_results = []

for split in split_values:

    classifier = DecisionTreeClassifier(
        criterion='entropy',
        max_depth=best_depth,
        min_samples_split=split,
        min_samples_leaf=1,
        random_state=42
    )

    scores = cross_val_score(
        classifier,
        x,
        y,
        cv=skfold,
        scoring='accuracy'
    )

    split_results.append([
        split,
        scores.mean()
    ])

split_df = pd.DataFrame(
    split_results,
    columns=['Min Samples Split', 'Mean Accuracy']
)

print("\nMIN_SAMPLES_SPLIT COMPARISON")
print(split_df)

best_split_index = split_df['Mean Accuracy'].idxmax()
best_split = split_values[best_split_index]

print("\nBest Min Samples Split:", best_split)
print("Best Mean Accuracy:",
      split_df.loc[best_split_index, 'Mean Accuracy'])


# =========================================================
# 3. MIN_SAMPLES_LEAF COMPARISON
# =========================================================

leaf_values = [1, 2, 5, 10, 15]

leaf_results = []

for leaf in leaf_values:

    classifier = DecisionTreeClassifier(
        criterion='entropy',
        max_depth=best_depth,
        min_samples_split=best_split,
        min_samples_leaf=leaf,
        random_state=42
    )

    scores = cross_val_score(
        classifier,
        x,
        y,
        cv=skfold,
        scoring='accuracy'
    )

    leaf_results.append([
        leaf,
        scores.mean()
    ])

leaf_df = pd.DataFrame(
    leaf_results,
    columns=['Min Samples Leaf', 'Mean Accuracy']
)

print("\nMIN_SAMPLES_LEAF COMPARISON")
print(leaf_df)

best_leaf_index = leaf_df['Mean Accuracy'].idxmax()
best_leaf = leaf_values[best_leaf_index]

print("\nBest Min Samples Leaf:", best_leaf)
print("Best Mean Accuracy:",
      leaf_df.loc[best_leaf_index, 'Mean Accuracy'])


# =========================================================
# FINAL DECISION TREE MODEL
# =========================================================

final_classifier = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=best_depth,
    min_samples_split=best_split,
    min_samples_leaf=best_leaf,
    random_state=42
)

final_scores = cross_val_score(
    final_classifier,
    x,
    y,
    cv=skfold,
    scoring='accuracy'
)

print("\nFINAL DECISION TREE MODEL")
print("Max Depth:", best_depth)
print("Min Samples Split:", best_split)
print("Min Samples Leaf:", best_leaf)
print("Fold Accuracies:", final_scores)
print("Mean Accuracy:", final_scores.mean())
```
