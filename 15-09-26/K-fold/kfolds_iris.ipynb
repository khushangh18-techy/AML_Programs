```python
# Remove unnecessary ID column
df = df.drop("Id", axis=1)

df.head()
```

```python
# Separate input and target
x = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
# Split data into training and test sets
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=45
)
```

```python
# Check training and testing sizes
print("X_train:", x_train.shape)
print("X_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

sc_x = StandardScaler()

x_train = sc_x.fit_transform(x_train)
x_test = sc_x.transform(x_test)
```

```python
# Create Decision Tree classifier
from sklearn.tree import DecisionTreeClassifier

classifier = DecisionTreeClassifier(
    criterion="entropy",
    random_state=0
)

classifier.fit(x_train, y_train)
```

```python
# Predict test data
y_pred = classifier.predict(x_test)

y_pred
```

```python
y_test
```

```python
# Calculate model accuracy
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
# Display confusion matrix
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
# Display classification report
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
# Predict a new flower
new_flower = np.array([[5.1, 3.5, 1.4, 0.2]])

prediction = classifier.predict(
    sc_x.transform(new_flower)
)

print("Predicted Species:", prediction[0])
```

```python
# Import StratifiedKFold for cross-validation
from sklearn.model_selection import StratifiedKFold, cross_val_score
```

```python
# Create 5-fold stratified cross-validation
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
# Create a new Decision Tree classifier using Entropy
cv_classifier = DecisionTreeClassifier(
    criterion="entropy",
    random_state=0
)
```

```python
# Perform 5-fold cross-validation using accuracy
cv_accuracy = cross_val_score(
    cv_classifier,
    x,
    y,
    cv=skf,
    scoring="accuracy"
)

print("Accuracy for each fold:")
print(cv_accuracy)

print("\nMean Cross-Validation Accuracy:")
print(cv_accuracy.mean())
```

```python
# Calculate precision for each fold
cv_precision = cross_val_score(
    cv_classifier,
    x,
    y,
    cv=skf,
    scoring="precision_macro"
)

print("Precision for each fold:")
print(cv_precision)

print("\nMean Precision:")
print(cv_precision.mean())
```

```python
# Calculate recall for each fold
cv_recall = cross_val_score(
    cv_classifier,
    x,
    y,
    cv=skf,
    scoring="recall_macro"
)

print("Recall for each fold:")
print(cv_recall)

print("\nMean Recall:")
print(cv_recall.mean())
```

```python
# Calculate F1 score for each fold
cv_f1 = cross_val_score(
    cv_classifier,
    x,
    y,
    cv=skf,
    scoring="f1_macro"
)

print("F1 Score for each fold:")
print(cv_f1)

print("\nMean F1 Score:")
print(cv_f1.mean())
```

```python
# Create a table showing performance for every fold
fold_results = pd.DataFrame({
    "Fold": [1, 2, 3, 4, 5],
    "Accuracy": cv_accuracy,
    "Precision": cv_precision,
    "Recall": cv_recall,
    "F1 Score": cv_f1
})

fold_results
```

```python
# Calculate the average performance across all 5 folds
mean_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],
    
    "Mean Score": [
        cv_accuracy.mean(),
        cv_precision.mean(),
        cv_recall.mean(),
        cv_f1.mean()
    ]
})

mean_results
```

```python
# Calculate standard deviation to measure variation between folds

print("Accuracy Mean:", cv_accuracy.mean())
print("Accuracy Std:", cv_accuracy.std())

print("\nPrecision Mean:", cv_precision.mean())
print("Precision Std:", cv_precision.std())

print("\nRecall Mean:", cv_recall.mean())
print("Recall Std:", cv_recall.std())

print("\nF1 Score Mean:", cv_f1.mean())
print("F1 Score Std:", cv_f1.std())
```
