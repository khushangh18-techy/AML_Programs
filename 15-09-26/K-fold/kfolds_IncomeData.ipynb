Testing for K-fold instead of train_test_split

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load the dataset
df = pd.read_csv(r"C:/Users/MSIS/Downloads/AML/MLLabprojects/data/income_data.csv")

# Display the data
print(df.head())

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check duplicate values
print("\nDuplicate values:", df.duplicated().sum())

# Select input and target
X = df[["income"]]
y = df["happiness"]

# K-Fold Cross Validation
kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Different polynomial degrees to compare
degrees = [1, 2, 3, 4, 5]

results = []

for degree in degrees:

    # Create Polynomial Regression model
    model = make_pipeline(
        PolynomialFeatures(degree=degree),
        LinearRegression()
    )

    # R2 scores using K-Fold
    r2_scores = cross_val_score(
        model,
        X,
        y,
        cv=kfold,
        scoring="r2"
    )

    # MSE scores using K-Fold
    mse_scores = -cross_val_score(
        model,
        X,
        y,
        cv=kfold,
        scoring="neg_mean_squared_error"
    )

    # Calculate average scores
    mean_r2 = r2_scores.mean()
    mean_mse = mse_scores.mean()
    mean_rmse = np.sqrt(mean_mse)

    results.append([
        degree,
        mean_r2,
        mean_mse,
        mean_rmse
    ])

# Create results table
results_df = pd.DataFrame(
    results,
    columns=[
        "Polynomial Degree",
        "Mean R2 Score",
        "Mean MSE",
        "Mean RMSE"
    ]
)

print("\nPolynomial Degree Comparison:")
print(results_df)

# Find the best degree based on R2
best_index = results_df["Mean R2 Score"].idxmax()
best_degree = degrees[best_index]

print("\nBest Polynomial Degree:", best_degree)

# Create final model using best degree
final_model = make_pipeline(
    PolynomialFeatures(degree=best_degree),
    LinearRegression()
)

# Train final model on the complete dataset
final_model.fit(X, y)

# Predictions
y_pred = final_model.predict(X)

# Final model evaluation
print("\nFinal Model Performance:")
print("MAE:", mean_absolute_error(y, y_pred))
print("MSE:", mean_squared_error(y, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y, y_pred)))
print("R2 Score:", r2_score(y, y_pred))

# Prediction for new income value
new_income = pd.DataFrame({
    "income": [5.2000]
})

prediction = final_model.predict(new_income)

print("\nPredicted Happiness for income = 5.2:",
      prediction[0])

# Plot actual data and best model
X_plot = np.linspace(
    X["income"].min(),
    X["income"].max(),
    300
).reshape(-1, 1)

y_plot = final_model.predict(X_plot)

plt.scatter(X, y)
plt.plot(X_plot, y_plot)

plt.xlabel("Income")
plt.ylabel("Happiness")
plt.title("Income vs Happiness - Polynomial Regression")
plt.grid(True)
plt.show()
```
